```python
%%writefile grid_env_det.py

states = ["S0", "S1", "S3", "S4", "G"]
actions = ["up", "down", "left", "right"]
gamma = 0.9

def transition(state, action):
    transitions = {
        "S0": {
            "up": "S0",
            "down": "S0",
            "left": "S0",
            "right": "S1"
        },
        "S1": {
            "up": "S1",
            "down": "S3",
            "left": "S0",
            "right": "S1"
        },
        "S3": {
            "up": "S1",
            "down": "S3",
            "left": "S3",
            "right": "S4"
        },
        "S4": {
            "up": "G",
            "down": "S4",
            "left": "S3",
            "right": "S4"
        }
    }
    return transitions[state][action]

def reward(state, action, next_state):
    if next_state == "G":
        return 10
    return -1
```

```python
from grid_env_det import transition, reward, gamma

policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}

def state_value_function(start_state):
    state = start_state
    total_return = 0
    discount = 1

    while state != "G":
        action = policy[state]
        next_state = transition(state, action)
        r = reward(state, action, next_state)
        total_return += discount * r
        discount *= gamma
        state = next_state

    return total_return

def action_value_function(state, action):
    total_return = 0
    discount = 1

    # Take the given action first
    next_state = transition(state, action)
    r = reward(state, action, next_state)
    total_return += discount * r
    discount *= gamma
    state = next_state

    # Follow the policy afterwards
    while state != "G":
        action = policy[state]
        next_state = transition(state, action)
        r = reward(state, action, next_state)
        total_return += discount * r
        discount *= gamma
        state = next_state

    return total_return

print("State Value of S0:", state_value_function("S0"))
print("Action Value of S0 for action Up:", action_value_function("S0", "up"))
```
