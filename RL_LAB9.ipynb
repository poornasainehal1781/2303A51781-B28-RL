```python
# REINFORCEMENT LEARNING LAB - 9
# Problem 1: Algebraic Bellman Equations
# Problem 2: Iterative Policy Evaluation

gamma = 0.9
theta = 0.000001

# ==========================================
# PROBLEM 1: ALGEBRAIC SOLUTION
# ==========================================

print("========== PROBLEM 1: ALGEBRAIC SOLUTION ==========")

# Solve the simultaneous equations directly
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

V0 = 71.819 / 0.9271
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_Down = -1 + gamma * V1
Q0_Right = -1
Q1_Right = 100
Q1_Up = -1 + gamma * V0

print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")
print(f"Q(S0, Down) = {Q0_Down:.2f}")
print(f"Q(S0, Right) = {Q0_Right:.2f}")
print(f"Q(S1, Right) = {Q1_Right:.2f}")
print(f"Q(S1, Up) = {Q1_Up:.2f}")

# Verify Bellman expectation equations
V0_check = 0.9 * Q0_Down + 0.1 * Q0_Right
V1_check = 0.9 * Q1_Right + 0.1 * Q1_Up

print("\nVerification:")
print(f"V(S0) = {V0_check:.2f}")
print(f"V(S1) = {V1_check:.2f}")


# ==========================================
# PROBLEM 2: ITERATIVE POLICY EVALUATION
# ==========================================

print("\n========== PROBLEM 2: ITERATIVE EVALUATION ==========")

# Initialize state values
V0 = 0.0
V1 = 0.0
iterations = 0

# Iterate until convergence
while True:
    # Calculate new values using previous values
    new_V0 = -1 + 0.81 * V1
    new_V1 = 89.9 + 0.09 * V0

    # Calculate maximum change
    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))

    # Update state values
    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

# Calculate action values using stochastic transitions
Q0_Down = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V1)
Q0_Right = -1
Q1_Right = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V0)
Q1_Up = -1 + gamma * V0

# Display results
print("Iterations:", iterations)
print(f"V(S0) = {V0:.6f}")
print(f"V(S1) = {V1:.6f}")

print("\nAction Values:")
print(f"Q(S0, Down) = {Q0_Down:.6f}")
print(f"Q(S0, Right) = {Q0_Right:.6f}")
print(f"Q(S1, Right) = {Q1_Right:.6f}")
print(f"Q(S1, Up) = {Q1_Up:.6f}")

# Check policy consistency
print("\nPolicy Consistency:")
print("V(S0) = Q(S0, Down):", abs(V0 - Q0_Down) < theta)
print("V(S1) = Q(S1, Right):", abs(V1 - Q1_Right) < theta)
```
