```python
import gymnasium as gym
import numpy as np

def random_agent(env_name, episodes=10):

    env = gym.make(env_name)

    rewards = []

    print("\n" + "=" * 50)
    print("Environment:", env_name)
    print("Action Space:", env.action_space)
    print("Observation Space:", env.observation_space)
    print("=" * 50)

    for episode in range(1, episodes + 1):

        state, info = env.reset()
        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        while not (terminated or truncated):

            action = env.action_space.sample()

            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1

            state = next_state

        rewards.append(total_reward)

        print(
            f"Episode {episode}: "
            f"Steps = {steps}, "
            f"Total Reward = {total_reward}"
        )

    average_reward = np.mean(rewards)
    total_reward_all_episodes = np.sum(rewards)

    print("\nResults")
    print("-" * 50)
    print("Total Reward:", total_reward_all_episodes)
    print("Average Reward:", average_reward)

    env.close()


random_agent("CartPole-v1", episodes=10)

random_agent("MountainCar-v0", episodes=10)

random_agent("FrozenLake-v1", episodes=10)
```
