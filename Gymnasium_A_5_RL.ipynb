```python
import math


machines = ['A', 'B', 'C']

N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': 0,
    'B': 0,
    'C': 0
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

UCB = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

N_play = 10

total_reward = 0



for play in range(1, N_play + 1):

    print("\n-----------------------------")
    print("Play:", play)
    print("-----------------------------")


    for machine in machines:

        # Make sure every machine is played at least once
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    selected_machine = max(machines, key=lambda x: UCB[x])

    print("Selected Machine:", selected_machine)




    reward = float(
        input("Enter reward obtained from machine "
              + selected_machine + ": ")
    )

    N[selected_machine] += 1

    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    total_reward += reward


    print("\nCurrent Values:")

    for machine in machines:
        print(
            machine,
            "N =", N[machine],
            "R =", R[machine],
            "Q =", round(Q[machine], 3),
            "UCB =", round(UCB[machine], 3)
            if UCB[machine] != float('inf')
            else "INF"
        )



print("\n\n====================================")
print("FINAL RESULTS")
print("====================================")

print("\nMachine\tN\tR\tQ-value\tUCB")


for machine in machines:

    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

    print(
        machine,
        "\t",
        N[machine],
        "\t",
        R[machine],
        "\t",
        round(Q[machine], 3),
        "\t",
        round(UCB[machine], 3)
        if UCB[machine] != float('inf')
        else "INF"
    )

print("\nTotal Reward =", total_reward)

# Best machine
best_machine = max(machines, key=lambda x: Q[x])

print("Best Machine =", best_machine)
print("Best Q-value =", round(Q[best_machine], 3))


```
